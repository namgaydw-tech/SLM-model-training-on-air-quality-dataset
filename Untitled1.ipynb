```python

import json
from pathlib import Path
from collections import Counter

DATASET = Path(
    r"D:\Design thinking LLM\air_quality_project\data\air_quality_llm_train.jsonl"
)

print("===== AIR QUALITY LLM DATASET =====")

records = []
invalid_lines = []

with DATASET.open("r", encoding="utf-8") as f:
    for line_number, line in enumerate(f, 1):
        if not line.strip():
            continue

        try:
            records.append(json.loads(line))
        except json.JSONDecodeError:
            invalid_lines.append(line_number)

print("Total records:", len(records))
print("Invalid JSON lines:", len(invalid_lines))

print("\n===== RECORD STRUCTURE =====")

if records:
    print("First record type:", type(records[0]).__name__)

    if isinstance(records[0], dict):
        print("Fields:", list(records[0].keys()))

print("\n===== FIRST 2 TRAINING EXAMPLES =====")

for i, record in enumerate(records[:2], 1):
    print(f"\n--- Example {i} ---")
    print(json.dumps(record, indent=2, ensure_ascii=False)[:2500])

print("\n===== DUPLICATE CHECK =====")

serialized = [
    json.dumps(r, sort_keys=True, ensure_ascii=False)
    for r in records
]

print("Exact duplicate records:",
      len(serialized) - len(set(serialized)))

print("\n===== RECORD FIELD DISTRIBUTION =====")

key_counts = Counter()

for record in records:
    if isinstance(record, dict):
        key_counts.update(record.keys())

for key, count in key_counts.items():
    print(f"{key}: {count}")

print("\nDataset inspection complete!")
```
